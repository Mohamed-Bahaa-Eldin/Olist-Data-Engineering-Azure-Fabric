```sql
--- dim products

drop table if exists dim_products;
CREATE TABLE dim_products (
    product_key                 BIGINT NOT NULL,
    product_id_bk               STRING not null,
    product_category_name       STRING,
    product_name_length         int , 
    product_description_length  int ,
    product_photo_qty           int ,
    weight_g                    INT,
    length_cm                   INT,
    height_cm                   INT,
    width_cm                    INT,
    created_at                  TIMESTAMP,
    updated_at                  TIMESTAMP,
    is_active                   BOOLEAN
) USING DELTA;
```

```sql
--------------------- dim_sellers 

drop table if exists dim_sellers;
create table dim_sellers(
 seller_key                 bigint not null,
 seller_id_bk               string not null,
 seller_zip_code_perfix     int , 
 seller_city                string , 
 seller_state               string , 
 geolocation_lat            float , 
 geolocation_lng            float ,
 created_at                 timestamp 
) using delta ;

```

```sql
-------------------- dim_customers

drop table if exists dim_customers  ; 
create table dim_customers(
customer_key                 bigint not null , 
customer_id_bk               string , 
customer_unique_id           string , 
customer_zip_code_prefix     int ,
customer_city                string ,
customer_state               string , 
geolocation_lat              float , 
geolocation_lng              float , 
created_at                   timestamp 
) ;



```

```sql

----------------- dim date

DROP TABLE IF EXISTS dim_date;
CREATE TABLE dim_date (
    date_key INT NOT NULL,
    full_date DATE,
    year INT,
    quarter STRING,
    month INT,
    month_name STRING,
    week_of_year INT,
    day_of_month INT,
    day_name STRING,
    is_weekend INT
);
```

```sql
----------------- dim time

DROP TABLE IF EXISTS dim_time;
CREATE TABLE dim_time (
    time_key INT NOT NULL,
    time_value VARCHAR(8), -- تم التعديل لـ 8 خانات ليتناسب مع HH:mm:ss
    hour INT,
    minute INT,
    second INT,             -- أضفنا عمود الثواني للكمال
    period STRING
);
```

```sql
------- order fact table

drop table if exists fact_orders;
create table fact_orders (
    order_key                           BIGINT not null,
    order_id_bk                         string , 
    customer_id_fk                      bigint,
    order_status                        string ,
    order_purchase_timestamp            timestamp ,
    order_purchase_date_fk              int,
    order_purchase_time_fk              int,
    order_aproved_at                    timestamp,
    order_approved_date_fk              int,
    order_approved_time_fk              int ,
    approved_performance_hours          int ,
    order_delivered_carrier_date_fk     int,
    order_delivered_carrier_time_fk     int ,
    order_delivered_customers_date_fk   int,
    order_delivered_customers_time_fk   int,
    order_estimated_deliverty_date      timestamp,
    deliverey_performance_days          int,
    total_process_days                  int ,
    red_falg                            boolean,
    review_score                        int ,
    review_comment_title                string, 
    review_comment_message              string,
    created_at                          timestamp 

)using delta;
```

```sql
----------- order item fact table 

drop table if exists fact_order_items;
create table fact_order_items(
    order_item_key          bigint not null , 
    order_id_fk             bigint, 
    product_id_fk           bigint,
    seller_id_fk            bigint,
    total_quantity          bigint ,
    unit_price              decimal(10,2),
    unit_freight            decimal(10,2),
    total_product_price      decimal(10,2),
    total_freight_price     decimal(10,2),
    total_order_value        decimal(10,2),
    shipping_limit_date     timestamp ,
    created_at               timestamp 
);

```

```sql
----------- order payments

drop table if exists fact_order_payments ;
create table fact_order_payments (

order_payment_key       bigint not null,
order_id_fk             bigint ,
payment_sequential      int ,
payment_type            string ,
payment_installments    int ,
payment_value           decimal(10,2),
created_at              timestamp

)using delta ;

```
